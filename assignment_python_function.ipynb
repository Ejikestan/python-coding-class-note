```python
# Call the function
sales_data = [5000, 8200, 3100, 9700, 4400, 7600]
def monthly_report(sales):
    total_sales = sum(sales)
    average_sales = total_sales / len(sales)
    
    best_month = sales.index(max(sales)) + 1
    worst_month = sales.index(min(sales)) + 1
    
    print(f"Total sales: {total_sales}")
    print(f"Average sales: {average_sales}")
    print(f"Best month: Month {best_month}")
    print(f"Worst month: Month {worst_month}")


monthly_report(sales_data)
```

```python
# Test the function
scores = [45, 72, 88, 55, 91, 60, 38, 79]

def count_category(data, low, high):
    count = 0
    for value in data:
        if low <= value <= high:
            count += 1
    return count


result = count_category(scores, 60, 80)

print(f"Number of scores between 60 and 80: {result}")
```

```python
def price_report(products, prices):
    # 1. Print each product with its price
    for product, price in zip(products, prices):
        print(f"{product}: {price}")
    
    # 2. Most expensive product
    max_price = max(prices)
    max_index = prices.index(max_price)
    print(f"Most expensive product: {products[max_index]}")
    
    # 3. Cheapest product
    min_price = min(prices)
    min_index = prices.index(min_price)
    print(f"Cheapest product: {products[min_index]}")
    
    # 4. Average price
    avg_price = sum(prices) / len(prices)
    print(f"Average price: {avg_price}")

# Call the function
products = ["Rice", "Oil", "Beans", "Sugar"]
prices = [1800, 2500, 1200, 900]

price_report(products, prices)
```
