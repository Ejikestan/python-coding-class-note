Any value in Python can be converted to a Boolean using the `bool` function.

Only the following values evaluate to `False` (they are often called *falsy* values):

1. The value `False` itself
2. The integer `0`
3. The float `0.0`
4. The empty value `None`
5. The empty text `""`
6. The empty list `[]`
7. The empty tuple `()`
8. The empty dictionary `{}`
9. The empty set `set()`
10. The empty range `range(0)`

Everything else evaluates to `True` (a value that evaluates to `True` is often called a *truthy* value).

```python
print(bool(False))
print(bool(0))
print(bool(None))
print(bool(""))
print(bool([]))
print(bool(()))
print(bool({}))
print(bool(set()))
```

# loop through this list of scores. 
# print each student's name and whether they passed(>=50) or fail
# scores = [("ali',72),('bola',45),('chidi',90),('dana',38),('ebuka',55)]

```python
scores = [('ali',72),('bola',45),('chidi',90),('dana',38),('ebuka',55)]
for name, score in scores:
    if score >= 50:
        print(name, "-passed")
    else:
        print(name, "-failed")
```

# ONLINE SHOPPING CART SIMULATION

# You have 8000 in your wallet, simulate purchasing items worth 1200, 3500, 800, 5000, and 600 from an online store.
# Return "Insufficient funds" when your wallet balance is less than the item price and complete the purchase,
# if the item price is less than or equal to the wallet balance, while printing the remaining balance after each transaction.

```python
balance = 8000

worth = [1200, 3500, 800, 5000, 600]
i = 0
while i < len(worth):
    amount = worth[i]
    if amount > balance:
        print(f'insufficient funds for this {amount}. (balance: {balance})')
else:
    balance = balance - amount
print(f'this purchase is completed {amount}. (balance: {balance})')
i = i + 1
```
